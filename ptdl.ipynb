```python
# ==========================================
# 1. KHAI BÁO CÁC THƯ VIỆN CẦN THIẾT
# ==========================================
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Cấu hình giao diện biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# ==========================================
# 2. TẢI TỆP DỮ LIỆU CỦA BẠN LÊN COLAB
# ==========================================
print("Hãy chọn tệp CSV từ máy tính của bạn:")
uploaded = files.upload()

# Tự động lấy tên tệp vừa tải lên
# The original code selected the first uploaded file, which was 'car.c45-names'.
# This file is a description file, not the actual data.
# The actual data file is 'car.data'.
file_name = 'car.data'
df = pd.read_csv(file_name, header=None)

# Thêm tên cột cho DataFrame để dễ hiểu và trực quan hóa hơn
column_names = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']
df.columns = column_names

# ==========================================
# 3. KIỂM TRA TỔNG QUAN DỮ LIỆU
# ==========================================
print("\n--- 5 dòng đầu tiên ---")
display(df.head())

print("\n--- Thông tin tập dữ liệu ---")
print(df.info())

print("\n--- Thống kê mô tả (Cột số) ---")
display(df.describe())

# Phân loại cột số và cột chuỗi/danh mục
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# ==========================================
# 4. TRỰC QUAN HÓA DỮ LIỆU
# ==========================================

# A. Phân phối các biến số (Histogram + KDE)
if num_cols:
    plt.figure(figsize=(12, 4 * len(num_cols)))
    for idx, col in enumerate(num_cols, 1):
        plt.subplot(len(num_cols), 1, idx)
        sns.histplot(df[col], kde=True, color='skyblue')
        plt.title(f'Phân phối của cột: {col}')
    plt.tight_layout()
    plt.show()

# B. Biểu đồ tần suất các biến danh mục (Countplot)
if cat_cols:
    for col in cat_cols:
        plt.figure(figsize=(8, 4))
        # Chỉ lấy tối đa top 10 giá trị phổ biến nhất để tránh rối mắt
        top_order = df[col].value_counts().iloc[:10].index
        sns.countplot(data=df, y=col, order=top_order, palette='viridis')
        plt.title(f'Tần suất các giá trị phổ biến nhất trong {col}')
        plt.tight_layout()
        plt.show()

# C. Ma trận tương quan giữa các cột số (Correlation Heatmap)
if len(num_cols) > 1:
    plt.figure(figsize=(8, 6))
    sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title('Ma trận tương quan giữa các biến số')
    plt.show()
```

### D. Biểu đồ thanh nhóm (Grouped Bar Chart): Mối quan hệ giữa hai biến danh mục

Chúng ta sẽ xem xét mối quan hệ giữa `class` (đánh giá xe) và `buying` (giá mua).

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='buying', hue='class', palette='viridis')
plt.title('Phân phối Class theo Buying Price')
plt.xlabel('Buying Price')
plt.ylabel('Count')
plt.legend(title='Class')
plt.tight_layout()
plt.show()
```

### E. Biểu đồ thanh xếp chồng (Stacked Bar Chart): Tỷ lệ giữa hai biến danh mục

Chúng ta sẽ xem xét tỷ lệ các loại `class` theo từng mức độ `safety`.

```python
contingency_table = pd.crosstab(df['safety'], df['class'], normalize='index')
contingency_table.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')
plt.title('Tỷ lệ Class theo Safety Level')
plt.xlabel('Safety Level')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(title='Class')
plt.tight_layout()
plt.show()
```

### F. Heatmap của Bảng Tần số chéo (Contingency Table Heatmap): Mối quan hệ giữa hai biến danh mục

Để trực quan hóa mối quan hệ giữa `buying` và `class` một cách định lượng hơn.

```python
contingency_table_buying_class = pd.crosstab(df['buying'], df['class'])

plt.figure(figsize=(10, 7))
sns.heatmap(contingency_table_buying_class, annot=True, cmap='YlGnBu', fmt='d', linewidths=.5)
plt.title('Heatmap của Bảng Tần số chéo: Buying vs Class')
plt.xlabel('Class')
plt.ylabel('Buying Price')
plt.tight_layout()
plt.show()
```
